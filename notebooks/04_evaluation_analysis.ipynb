### Part 2: Comparative Analysis

| Embedding | Strengths | Weaknesses | Best For |
| :--- | :--- | :--- | :--- |
| **TF-IDF** | High keyword discrimination, efficient | No semantic understanding | Classification tasks with distinct vocabulary |
| **Word2Vec** | Captures semantics and context | Requires large data, slower training | Semantic search, clustering |
| **FastText** | Handles Out-of-Vocabulary (OOV) words | Higher memory usage | Tasks with complex/rare legal terms |
| **Doc2Vec** | Holistic document representation | Harder to tune, can be unstable | Document similarity, deduplication |

### Key Observations
1. **Sparse Models Excel:** TF-IDF paired with Random Forest achieved the highest accuracy. This suggests that for this legal dataset, identifying specific keywords is more critical than complex semantic relationships.
2. **Dense Models Robustness:** While slightly lower in accuracy, Word2Vec Skip-Gram showed strong performance with Logistic Regression, indicating that dense representations are linearly separable.
3. **Computational Efficiency:** TF-IDF and BoW are significantly faster to train compared to Averaged Word Embeddings from dense models.
